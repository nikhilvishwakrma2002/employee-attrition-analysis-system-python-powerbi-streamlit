```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Main Attrition
sns.countplot(x='Attrition', data=df)
plt.title("Employee Attrition Distribution")
plt.savefig("../Reports/attrition_distribution.png")
plt.show()
```

```python
# Dipartment wise
plt.figure(figsize=(8,5))

sns.countplot(
    x='Department',
    hue='Attrition',
    data=df
)

plt.xticks(rotation=15)
plt.title("Attrition by Department")
plt.savefig("../Reports/department_attrition.png")
plt.show()
```

```python
sns.countplot(
    x='OverTime',
    hue='Attrition',
    data=df
)

plt.title("Overtime vs Attrition")
plt.savefig("../Reports/overtime_vs_attrition.png")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='Attrition',
    y='MonthlyIncome',
    data=df
)

plt.title("Salary vs Attrition")
plt.savefig("../Reports/salary_vs_attrition.png")
plt.show()
```

```python
sns.countplot(
    x='JobSatisfaction',
    hue='Attrition',
    data=df
)

plt.title("Job Satisfaction vs Attrition")
plt.savefig("../Reports/job_satisfaction.png")
plt.show()
```

```python
eda_df = df.copy()

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in eda_df.select_dtypes(include='object').columns:
    eda_df[col] = le.fit_transform(eda_df[col])
    
plt.figure(figsize=(18,12))

sns.heatmap(
    eda_df.corr(),
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.savefig("../Reports/correlation_heatmap.png")
plt.show()
```

```python
'''
1. Employees doing overtime show higher attrition.

2. Sales department has highest employee turnover.

3. Lower salary employees are more likely to leave.

4. Low job satisfaction strongly impacts attrition.
'''
```
